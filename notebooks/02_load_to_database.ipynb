connect

```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus
from getpass import getpass

password = quote_plus(getpass("PostgreSQL password: "))
engine = create_engine(
    f"postgresql+psycopg2://postgres:{password}@localhost:5432/shopsense_db"
)
```

load the CSV and write it to the database

```python
df = pd.read_csv("../data/shopping_trends_cleaned.csv")
df.to_sql("customer", engine, if_exists="replace", index=False)
```

Verify it worked

```python
pd.read_sql("SELECT COUNT(*) FROM customer", engine)
```
